# Adaptive PDR — Interactive Quick Demo

This notebook runs the repository's synthetic end-to-end Pedestrian Dead Reckoning pipeline. It is designed as a quick interactive entry point before replacing the synthetic data with real smartphone IMU recordings.

```python
import os, pathlib, subprocess, sys
repo = pathlib.Path('/content/adaptive-pdr-research')
if not repo.exists():
    subprocess.run(['git','clone','https://github.com/Akashperla/adaptive-pdr-research.git', str(repo)], check=True)
os.chdir(repo)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', 'requirements.txt'], check=True)
print('Repository ready:', repo)
```

## Run the synthetic research baseline
The demo generates IMU data, trains the movement classifier baseline, runs Adaptive PDR, evaluates the trajectory, and saves output files.

```python
subprocess.run([sys.executable, 'scripts/run_demo.py'], check=True)
```
